```python
from enderscope import ScanPatterns

origin = [25,100] # coordinates of first position
steps = [38,-25] # x and y shifts to next item
positions = ScanPatterns.raster(4,3) * steps + origin
ScanPatterns.plot_path(positions, field=(40,30),title='raster scan')
```

```python
positions = ScanPatterns.snake(4,4) * [33.33,-25] + [50,100]
ScanPatterns.plot_path(positions, title='snake scan')
```

```python
positions = ScanPatterns.random(20)
ScanPatterns.plot_path(positions, title='random scan')
```

```python
positions = ScanPatterns.spiral(16) * [30,20]  + [100,75]
ScanPatterns.plot_path(positions,field=(34,24),title='spiral scan')
```

## Maintaining multiwell plate templates

```python
import ipywidgets as widgets
from IPython.display import display

# Dictionary with plates characteristics
PLATES = {
    "12 well": {"rows": 3, "cols": 4, "pitch": 25.0},
    "24 well": {"rows": 4, "cols": 6, "pitch": 19.3},
    "48 well": {"rows": 6, "cols": 8, "pitch": 13.0},
    "96 well": {"rows": 8, "cols": 12, "pitch": 9.0},
    "384 well": {"rows": 16, "cols": 24, "pitch": 4.50},
}

plate_selector = widgets.Dropdown(
    options=list(PLATES.keys()),
    description="Select plate model: "
)

display(plate_selector)
```

```python
# get selected model
plate=PLATES[plate_selector.value]
print( plate)
```

```python
# use it in a snake pattern definition
origin = [50,100]
steps = [(plate['pitch']),-(plate['pitch'])] # x and y offsets to next item
positions = ScanPatterns.snake(plate['cols'],plate['rows']) * steps + origin
ScanPatterns.plot_path(positions,field=(8,8),title='multiwell plate snake scan')
```
