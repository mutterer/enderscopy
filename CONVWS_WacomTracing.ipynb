```python

import xml.etree.ElementTree as ET

def parseXML(xmlfile):

    tree = ET.parse(xmlfile)

    root = tree.getroot()

    values = []
    for item in root:

        for child in item:
            values.append(list(child.attrib.values()))
 
 
    # return news items list
    return values
parseXML("strokes.xml")

def dimension_mapping(x,y,pressure):
    max_ender=220.0
    max_if=800
    max_pressure = 2047
    max_z=100
    min_z=1

    x_map = float(x) * max_ender/max_if
    y_map = float(y) * max_ender/max_if 
    z_map = min_z + (max_z -min_z) * float(pressure)/max_pressure

    return x_map,y_map,z_map

def coord_map_xml(xmlfile):
    values = parseXML(xmlfile)
    coords = []
    for list in values:
        x_coord,y_coord,z_coord = dimension_mapping(list[0], list[1],list[2])
        coords.append([x_coord,y_coord,z_coord])

    return(coords)

coords = coord_map_xml("stroke_1.xml")
print(coords)

for coord in coords:
    s.move_absolute(coord[0],coord[1],coord[2])
```
